```python
from google.colab import files
import pandas as pd

# This will open a file picker dialog for you to upload the file
uploaded = files.upload()

# Assuming the user uploads a file, we'll get its name
if uploaded:
    # The uploaded dictionary keys are the filenames
    file_name = list(uploaded.keys())[0]
    print(f"File '{file_name}' uploaded successfully. Loading into DataFrame...")

    # Load the uploaded CSV file into a pandas DataFrame
    # If your file has a different name than expected (e.g., 'zomato.csv' instead of 'Zomato-data-.csv'),
    # you might need to adjust the file_name variable or directly use its name.
    df = pd.read_csv(file_name)
    print("DataFrame loaded. Displaying first 5 rows:")
    display(df.head())
else:
    print("No file was uploaded. Please try again.")

# Now, apply the subsequent operations (renaming and dropping 'votes' column)
# if the DataFrame 'df' was successfully loaded.
if 'df' in locals() and not df.empty:
    # Rename the 'name' column to 'Rest_Name' permanently
    df.rename(columns={"name": "Rest_Name"}, inplace=True)
    print("\nColumn 'name' renamed to 'Rest_Name':")
    display(df.head())

    # Permanently drop the 'votes' column
    if 'votes' in df.columns:
        df.drop("votes", axis=1, inplace=True)
        print("\nColumn 'votes' dropped:")
        display(df.head())
    else:
        print("\n'votes' column not found, perhaps already dropped or not present in the dataset.")
else:
    print("DataFrame 'df' was not loaded, skipping further operations.")
```
